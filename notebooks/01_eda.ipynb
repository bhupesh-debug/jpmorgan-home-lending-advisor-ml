# 01 — Exploratory Data Analysis

**All data is synthetic.** No real borrower data, no real PII, no real JPMorgan Chase data.

Three datasets: monthly pipeline (branch x product x month), borrowers (synthetic applicants), customers (existing bank customers).

```python
import pandas as pd
pipe = pd.read_csv('data/processed/monthly_pipeline.csv', parse_dates=['month'])
bor = pd.read_csv('data/processed/borrowers.csv', parse_dates=['application_date'])
cust = pd.read_csv('data/processed/customers.csv')
print(pipe.shape, bor.shape, cust.shape)
print(pipe.describe().T)
print(bor['affordability_band'].value_counts())
print(bor['product_fit'].value_counts())
print(cust['mortgage_need'].mean())
```

## Funnel check

Pre-approvals <= applications <= in-underwriting <= approved <= funded, every row. Rates stay in [5, 10]; DTI in (0, 1.2]; credit scores in [300, 850].

```python
assert (pipe['pre_approvals'] <= pipe['applications']).all()
assert (pipe['funded_count'] <= pipe['approved']).all()
assert bor['back_dti'].between(0, 1.2).all()
print('funnel + bounds OK')
```
